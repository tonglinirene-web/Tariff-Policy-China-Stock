# 02 Event Study

Estimate market-model abnormal returns and produce AAR, CAR, and CAAR tables for the three tariff events.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.config import DATA_PROCESSED, OUTPUT_TABLES, EVENTS, EVENT_WINDOWS, ESTIMATION_LENGTH, VALIDATION_TARGETS
from src.event_study import run_event_study
from src.statistical_tests import aar_tests, car_tests, validate_caar_targets
```

```python
returns = pd.read_csv(DATA_PROCESSED / "returns_cleaned.csv", parse_dates=["date"])
results = run_event_study(returns, EVENTS, EVENT_WINDOWS, estimation_length=ESTIMATION_LENGTH)

for name, table in results.items():
    table.to_csv(OUTPUT_TABLES / f"{name}.csv", index=False)

aar_tests(results["abnormal_returns"]).to_csv(OUTPUT_TABLES / "aar_tests.csv", index=False)
car_tests(results["car"]).to_csv(OUTPUT_TABLES / "car_tests.csv", index=False)
validation = validate_caar_targets(results["aar_caar"], VALIDATION_TARGETS)
validation.to_csv(OUTPUT_TABLES / "chapter4_validation.csv", index=False)
validation
```
